# 1. Growth Of Function
Time Complexity/Order of Growth defines the amount of time taken by any program with respect to the size of the input. Time Complexity specifies how the program would behave as the order of size of input is increased. So, Time Complexity is just a function of size of its input. We will go through some of basic and most common time complexities such as:

- *Constant Time Complexity O(1) : constant running time*
- *Linear Time Complexity O(n) : linear running time*
- *Logarithmic Time Complexity O(log n) : logarithmic running time*
- *Log-Linear Time Complexity O(n log n) : log-­linear running time*
- *Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)*
- *Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)*

## Constant time Complexity:O(1)
An algorithm is said to have constant time with order O (1) when it is not dependent on the input size n. Irrespective of the input size n, the runtime will always be the same. See example below

```python
# Load the Python libraries
import matplotlib.pyplot as plt
```

```python
# Prob_1

def add_two_numbers(a, b):
    return a + b

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    add_two_numbers(n, n + 1)
    time_values.append(1)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(1) Constant Time Complexity for Addition")
plt.show()
```

```python
#prob 2
import matplotlib.pyplot as plt

def is_even_or_odd(n):
    # This operation takes the same amount of time regardless of n's value
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

n_values = list(range(1, 1001, 10))
time_values = []

for n in n_values:
    is_even_or_odd(n)
    # We record a constant value of 1 to represent a fixed execution step
    time_values.append(1)

plt.figure(figsize=(10, 5))
plt.plot(n_values, time_values, color='blue', linewidth=2)
plt.ylim(0, 2)  # Setting y-axis to show the flat line clearly
plt.xlabel("Input value (n)")
plt.ylabel("Operations / Steps")
plt.title("O(1) Constant Time Complexity (Even/Odd Check)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

#Linear Time Complexity O(n) : linear running time
- *An algorithm is said to have a linear time complexity when the running time increases linearly with the length of the input. When the function involves checking all the values in input data, with this order O(n).*

```python
def linear_time_sum(numbers):
    total = 0
    for value in numbers:
        total += value
    return total

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    data = list(range(n))
    linear_time_sum(data)
    time_values.append(n)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(n) Linear Time Complexity")
plt.show()
```

#Logarithmic Time Complexity O(log n) : logarithmic running time
- *An algorithm is said to have a logarithmic time complexity when it reduces the size of the input data in each step. This indicates that the number of operations is not the same as the input size. The number of operations gets reduced as the input size increases. Algorithms are found in binary tree or binary search functions. This involves the search of a given value in an array by splitting the array into two and starting searching in one split. This ensures the operation is not done on every element of the data.*

```python
import matplotlib.pyplot as plt
import math

def logarithmic_steps(n):
    # This simulates the number of divisions in a binary search
    if n <= 0:
        return 0
    return math.log2(n)

# Testing for a wide range of input sizes to see the "flattening" curve
n_values = list(range(1, 1001))
time_values = []

for n in n_values:
    # We use log2(n) to represent the number of steps taken
    steps = logarithmic_steps(n)
    time_values.append(steps)

plt.figure(figsize=(10, 5))
plt.plot(n_values, time_values, color='red', linewidth=2)
plt.xlabel("Input size (n)")
plt.ylabel("Operations / Steps (log2 n)")
plt.title("O(log n) Logarithmic Time Complexity (Binary Search)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

#*Log-Linear Time Complexity O(n log n) : log-­linear running time*


* When we call a Logarithmic Time Algorithm inside a loop, it would
result into a Log-Linear Time Complexity program.




```python
#Prob_5 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example.
import matplotlib.pyplot as plt
import numpy as np

def log_linear_work(n):
    if n <= 1:
        return 0
    # Formula for n log n complexity
    return n * np.log2(n)

n_values = list(range(1, 1001, 10))
time_values = []

for n in n_values:
    # Calculating the theoretical number of operations
    steps = log_linear_work(n)
    time_values.append(steps)

plt.figure(figsize=(10, 5))
plt.plot(n_values, time_values, color='orange', linewidth=2)
plt.xlabel("Input size (n)")
plt.ylabel("Operations (n * log2 n)")
plt.title("O(n log n) Log-Linear Time Complexity (Merge Sort / QuickSort)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)

* When the computation time increases as function of N raised to some power, N being the size of input. Such a code has Polynomial Time complexity.




```python
# Prob_6 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example
import matplotlib.pyplot as plt

def quadratic_time_pairs(n):
    # Simulating a nested loop: for i in n: for j in n:
    # The total operations are n * n
    return n ** 2

n_values = list(range(1, 101)) # Smaller range because n^2 grows fast
time_values = []

for n in n_values:
    # Calculating the theoretical steps
    steps = quadratic_time_pairs(n)
    time_values.append(steps)

plt.figure(figsize=(10, 5))
plt.plot(n_values, time_values, color='purple', linewidth=2)
plt.xlabel("Input size (n)")
plt.ylabel("Operations (n^2)")
plt.title("O(n^2) Polynomial (Quadratic) Time Complexity")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

#Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)

* When the computation time of a code increases as function of X^N, N being the size of input. Such a code has Polynomial Time Complexity.




```python
# Prob_7 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example
import matplotlib.pyplot as plt
import numpy as np
import timeit

def simulated_merge_sort_work(n):
    if n <= 1:
        return 1
    # n log n represents the recursive splitting (log n)
    # and the linear merging (n) at each level
    return n * np.log2(n)

n_values = np.linspace(1, 1000, 50)
time_values = [simulated_merge_sort_work(n) for n in n_values]

plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values, color='teal', marker='v', markevery=5, label="O(n log n)")

# Adding a Linear line for comparison
plt.plot(n_values, n_values * 5, color='gray', linestyle='--', label="O(n) for comparison")

plt.title("O(n log n) Log-Linear Complexity", fontsize=14)
plt.xlabel("Input Size (n)")
plt.ylabel("Computational Work")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import math

def quick_sort(arr):
    # Base case: arrays with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr, 0

    # Selecting the pivot (middle element)
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    # Recursively sort and track depth
    left_sorted, left_depth = quick_sort(left)
    right_sorted, right_depth = quick_sort(right)

    # The depth of this branch is 1 + the deepest sub-branch
    current_depth = 1 + max(left_depth, right_depth)

    return left_sorted + middle + right_sorted, current_depth

# --- Visualization of Logarithmic Depth ---

n_values = [10, 50, 100, 500, 1000, 5000, 10000]
actual_depths = []
theoretical_log = []

for n in n_values:
    # Create a random array
    test_arr = np.random.permutation(n).tolist()

    # Perform sort and get recursion depth
    _, depth = quick_sort(test_arr)
    actual_depths.append(depth)

    # Calculate theoretical log2(n) for comparison
    theoretical_log.append(math.log2(n))

plt.figure(figsize=(10, 6))
plt.plot(n_values, actual_depths, 'o-', label="Measured Recursion Depth")
plt.plot(n_values, theoretical_log, 'r--', label="Theoretical O(log n)")

plt.title("QuickSort Recursion Depth: Demonstrating Logarithmic Scaling")
plt.xlabel("Input Size (n)")
plt.ylabel("Tree Depth (Recursive Levels)")
plt.xscale('log') # Log scale helps visualize the linear relationship of log growth
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.show()
```
